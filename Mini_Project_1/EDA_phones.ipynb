```python
df = df.drop_duplicates()
```

```python
# univariate analysis

numerical_columns = df.select_dtypes(include=['int64', 'float64']).columns

categorical_columns = df.select_dtypes(include=['object']).columns

numerical_columns
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
for col in numerical_columns:
  plt.figure(figsize=(8, 6))
  sns.histplot(df[col], kde=True)
  plt.title(f'Distribution of {col}')
  plt.show()
```

```python
#Bivariate

for col in categorical_columns:

    plt.figure(figsize=(8, 6))

    sns.countplot(data=df, x=col)

    plt.title(f'Distribution of {col}')

    plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data = df, x="Battery_Backup_Hrs",y="Price_USD")
plt.title("Battery Backup vs Price")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df,x="Brand",y="Price_USD")
plt.title("Brand vs Price")
plt.show()
```

```python
#visualization
sns.pairplot(df)
plt.show()
```

```python
plt.figure(figsize=(10,8))
sns.heatmap(df.corr(numeric_only=True),annot=True, cmap="coolwarm",linewidth=0.5,fmt=".2f")
plt.title("Correlation mattrix")
plt.show()
```
